```python
# Cell 2 — Load dataset
# Read the CSV file into a DataFrame (pandas table structure)
# encoding="latin-1" is needed because the file contains special characters
df = pd.read_csv(r"C:\Users\admin\Desktop\Data Analyst\DataSet\Sample - Superstore.csv", encoding="latin-1")

# Print the dimensions of the dataset (rows, columns)
print(df.shape)

# Display the first 5 rows to get a quick look at the data
df.head()
```

```python
# Cell 4 — Business question
# Which is the top-selling category?
# Group rows by Category, sum the Sales column, and sort from highest to lowest
df.groupby("Category")["Sales"].sum().sort_values(ascending=False)
```
